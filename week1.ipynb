```python
import numpy as np

# 1. Load the raw dataset
def load_data():
    data = [
        {"name": "Abebe", "score": "85"},
        {"name": "Sara", "score": "90"},
        {"name": "Meron", "score": ""},
        {"name": "Dawit", "score": "75"},
        {"name": "Hana", "score": "invalid"},
        {"name": "Yonas", "score": "95"}
    ]
    return data


# 2. Clean the dataset
def clean_data(data):
    cleaned_data = []

    for row in data:
        try:
            name = row["name"].strip()
            score = float(row["score"])

            if name and 0 <= score <= 100:
                cleaned_data.append({
                    "name": name,
                    "score": score
                })

        except (ValueError, TypeError):
            continue

    return cleaned_data


# 3. Process the cleaned data using NumPy
def process_data(data):
    scores = np.array([row["score"] for row in data])

    print("Number of valid records:", len(scores))
    print("Average score:", round(np.mean(scores), 2))
    print("Highest score:", np.max(scores))
    print("Lowest score:", np.min(scores))


# 4. Display the cleaned dataset
def display_data(data):
    print("\nCleaned Dataset")
    print("----------------")

    for row in data:
        print(f"{row['name']}: {row['score']}")


# 5. Run the application
raw_data = load_data()
cleaned_data = clean_data(raw_data)

display_data(cleaned_data)
process_data(cleaned_data)
```
